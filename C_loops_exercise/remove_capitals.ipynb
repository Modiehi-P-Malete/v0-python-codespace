Write a function `remove_capitals(text)` that returns a new string with all
capital letters removed.




```python
def remove_capitals(text):
    result = ""
    for char in text:
        if not char.isupper():
            result += char
    return result


print(remove_capitals("fOrEver"))
print(remove_capitals("raiNCoat"))
print(remove_capitals("cElLAr Door"))

```
