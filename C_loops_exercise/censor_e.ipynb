Write a function `censor_e(text)` that returns a new string where all 'e' characters
are replaced with '*'.





```python
def censor_e(text):
    result = ""
    for char in text:
        if char == "e":
            result += "*"
        else:
            result += char
    return result


print(censor_e("speedy"))  #-> 'sp**dy'
print(censor_e("pending")) #-> 'p*nding'
print(censor_e("scene"))   #-> 'sc*n*'
print(censor_e("heat"))    #-> 'h*at'
```
